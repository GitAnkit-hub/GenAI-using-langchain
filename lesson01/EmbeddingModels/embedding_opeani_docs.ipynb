```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()
```

```python
embedding = OpenAIEmbeddings(model='text-embedding-3-small', dimensions=32)
documents = [
    "Delhi is the capital of the india"
    "Kolkata is the capital of the india"
    "Pairs is the capital of the france"
]

res = embedding.embed_documents(documents)
# res = embedding.embed_query("My name is ankit agrawal")

print(len(res))
print(str(res))
```
